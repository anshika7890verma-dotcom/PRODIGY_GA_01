```python
import random
import re
from collections import defaultdict, Counter

print("All libraries imported successfully!")
```

```python
text = """
Artificial intelligence is changing the world.
Artificial intelligence helps people solve problems.
Machine learning is a part of artificial intelligence.
Machine learning allows computers to learn from data.
Python is a popular programming language.
Python is widely used in artificial intelligence.
Generative AI can create text and images.
Generative AI is changing modern technology.
Technology makes our lives easier.
Technology helps people communicate.
"""

print("Dataset loaded successfully!")
print(text)
```

```python
# Convert text to lowercase
text = text.lower()

# Extract individual words
words = re.findall(r"\b[\w']+\b", text)

print("Preprocessed Words:")
print(words)

print("\nTotal number of words:", len(words))
```

```python
# Create a Markov Chain model
markov_chain = defaultdict(Counter)

# Learn relationships between consecutive words
for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]

    markov_chain[current_word][next_word] += 1

print("Markov Chain Model Created Successfully!")

# Display some learned patterns
for word, next_words in list(markov_chain.items())[:10]:
    print(word, "->", dict(next_words))
```

```python
def generate_text(start_word, length=20):

    current_word = start_word.lower()
    generated_words = [current_word]

    for _ in range(length - 1):

        if current_word not in markov_chain:
            break

        possible_words = list(markov_chain[current_word].keys())
        probabilities = list(markov_chain[current_word].values())

        next_word = random.choices(
            possible_words,
            weights=probabilities
        )[0]

        generated_words.append(next_word)
        current_word = next_word

    return " ".join(generated_words)


print("Generated Text:")
print(generate_text("artificial", 20))
```

```python
print("===================================")
print("     MARKOV CHAIN TEXT GENERATOR")
print("===================================")

start_word = input("Enter a starting word: ")

length = int(input("Enter the number of words to generate: "))

generated_text = generate_text(start_word, length)

print("\nGenerated Text:")
print(generated_text)
```

```python
test_words = ["artificial", "machine", "python", "technology", "generative"]

print("MARKOV CHAIN MODEL TESTING\n")

for word in test_words:
    print("Starting Word:", word)
    print("Generated Text:", generate_text(word, 10))
    print("-" * 50)
```
